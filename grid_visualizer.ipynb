```python
import numpy as np
import random
import matplotlib.pyplot as plt
from collections import deque
from grid_generator import generate_grid
```

```python
def visualize_grid(grid, start, goal, title="Random Grid"):

    rows, cols = grid.shape
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.imshow(grid, cmap="Greys", vmin=0, vmax=1)

    ax.scatter(start[1], start[0],color="green", s=250, marker="o",label="Start", edgecolors="black")
    ax.scatter(goal[1], goal[0],color="red", s=250, marker="*",label="Goal", edgecolors="black")

    ax.set_xticks(np.arange(-0.5, cols, 1))
    ax.set_yticks(np.arange(-0.5, rows, 1))

    ax.grid(color="black", linewidth=0.5)

    ax.set_xticklabels([])
    ax.set_yticklabels([])

    ax.set_title(title)
    ax.legend(loc="upper right")

    plt.show()
```

```python
grid, start, goal = generate_grid(
    rows=10,
    cols=10,
    obstacle_prob=0.25
)

print("Grid:\n", grid)
print("Start:", start)
print("Goal:", goal)

visualize_grid(grid, start, goal)
```
