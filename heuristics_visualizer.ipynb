```python
import numpy as np
import matplotlib.pyplot as plt
import heapq
import time
import pandas as pd
from heuristics import manhattan_distance, euclidean_distance, chebyshev_distance, heuristics
```

```python
grid = np.loadtxt("results/grid.csv", delimiter=",", dtype=int)
start = tuple(np.loadtxt("results/start.csv", delimiter=",", dtype=int))
goal = tuple(np.loadtxt("results/goal.csv", delimiter=",", dtype=int))

print("Grid shape:", grid.shape)
print("Start:", start)
print("Goal:", goal)
```

```python
current = (2, 3)
goal = (7, 8)

print("Manhattan Distance:", manhattan_distance(current, goal))
print("Euclidean Distance:", euclidean_distance(current, goal))
print("Chebyshev Distance:", chebyshev_distance(current, goal))
```

```python
def visualize_heuristics(grid, goal):

    rows, cols = grid.shape

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    for ax, (name, heuristic) in zip(axes, heuristics.items()):

        values = np.zeros((rows, cols))

        for r in range(rows):
            for c in range(cols):
                values[r, c] = heuristic((r, c), goal)

        im = ax.imshow(values, cmap="viridis")

        ax.scatter(
            goal[1], goal[0],
            color="red", marker="*", s=200,
            edgecolors="white"
        )

        ax.set_title(name)
        ax.set_xticks(np.arange(-0.5, cols, 1))
        ax.set_yticks(np.arange(-0.5, rows, 1))
        ax.grid(color="white", linewidth=0.3)

        fig.colorbar(im, ax=ax, label="Heuristic Value")

    plt.tight_layout()
    plt.show()
```

```python
visualize_heuristics(grid, goal)
```
