```python
import numpy as np
import matplotlib.pyplot as plt
import heapq
import time
import pandas as pd
```

```python
grid = np.loadtxt("results/grid.csv", delimiter=",", dtype=int)

start = tuple(np.loadtxt("results/start.csv", delimiter=",", dtype=int))
goal = tuple(np.loadtxt("results/goal.csv", delimiter=",", dtype=int))

print("Grid shape:", grid.shape)
print("Start:", start)
print("Goal:", goal)
```

```python
def manhattan_distance(current, goal):
    return abs(current[0] - goal[0]) + abs(current[1] - goal[1])
```

```python
def euclidean_distance(current, goal):
    return np.sqrt((current[0] - goal[0])**2 +(current[1] - goal[1])**2)
```

```python
def chebyshev_distance(current, goal):
    return max(abs(current[0] - goal[0]),abs(current[1] - goal[1]))
```

```python
current = (2, 3)
goal = (7, 8)

print("Manhattan Distance:", manhattan_distance(current, goal))
print("Euclidean Distance:", euclidean_distance(current, goal))
print("Chebyshev Distance:", chebyshev_distance(current, goal))
```
