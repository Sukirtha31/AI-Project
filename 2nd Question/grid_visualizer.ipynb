```python
def visualize_grid(grid, start, goal, title="Random Grid"):

    rows, cols = grid.shape
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.imshow(grid, cmap="Greys", vmin=0, vmax=1)
    colors = plt.cm.tab10(np.linspace(0, 1, len(start)))

    for i in range(len(start)):
        colour = colors[i]
        ax.scatter(start[i][1], start[i][0], color=colour, s=250, marker="o", label=f"Start {i+1}", edgecolors="black")
        ax.scatter(goal[i][1], goal[i][0], color=colour, s=250, marker="*", label=f"Goal {i+1}", edgecolors="black")

    ax.set_xticks(np.arange(-0.5, cols, 1))
    ax.set_yticks(np.arange(-0.5, rows, 1))

    ax.grid(color="black", linewidth=0.5)

    ax.set_xticklabels([])
    ax.set_yticklabels([])

    ax.set_title(title)
    ax.legend(loc="upper right")

    plt.show()
```

```python
grid, start, goal = generate_grid(
    rows=10,
    cols=10,
    obstacle_prob=0.25,
    no_of_robots=4
)

print("Grid:\n", grid)
print("Start:", start)
print("Goal:", goal)

visualize_grid(grid, start, goal)
```
